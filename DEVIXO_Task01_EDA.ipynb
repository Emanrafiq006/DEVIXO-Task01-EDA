## 1. Import libraries and load the dataset

The dataset should be placed in `data/supermarket_sales.csv`. If it is not available, download the Kaggle dataset linked in the README.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

df = pd.read_csv("../data/supermarket_sales.csv")
df.head()
```

```python
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
print("\nData types:")
print(df.dtypes)
```

```python
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
categorical_cols = df.select_dtypes(include="object").columns.tolist()
print("Numerical columns:", numeric_cols)
print("Categorical columns:", categorical_cols)

print("\nStatistical summary:")
display(df.describe())
```

## 2. Data cleaning

I checked missing values, duplicates, data types and basic validity. The original dataset is already fairly clean, so I avoid unnecessary changes. Dates and times are converted to useful datetime/time features for analysis.

```python
print("Missing values before cleaning:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# Remove exact duplicate rows if any are present
df = df.drop_duplicates().copy()

# Convert date/time fields
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df["Time"] = pd.to_datetime(df["Time"], format="%H:%M", errors="coerce").dt.time

# Create analysis-friendly features
df["Month"] = df["Date"].dt.month_name()
df["Day"] = df["Date"].dt.day_name()
df["Hour"] = pd.to_datetime(df["Time"].astype(str)).dt.hour

# Basic validity checks
print("Invalid quantity values:", (df["Quantity"] <= 0).sum())
print("Invalid total values:", (df["Total"] <= 0).sum())
print("Invalid ratings:", ((df["Rating"] < 1) | (df["Rating"] > 10)).sum())

print("\nShape after cleaning:", df.shape)
print("Missing values after cleaning:")
print(df.isnull().sum())
```

### Cleaning decisions
- No values are filled blindly because the original transaction data does not require artificial replacement.
- Exact duplicate rows are removed.
- Date and time are converted into analysis-friendly formats.
- Month, day and hour are derived from the transaction date/time to support trend analysis.
- Basic range checks are used for quantity, total and customer rating.

## 3. Feature analysis

```python
print("Average unit price:", round(df["Unit price"].mean(), 2))
print("Average quantity:", round(df["Quantity"].mean(), 2))
print("Average transaction value:", round(df["Total"].mean(), 2))
print("Average gross income:", round(df["gross income"].mean(), 2))
print("Average rating:", round(df["Rating"].mean(), 2))

print("\nSales by product line:")
display(df.groupby("Product line")["Total"].sum().sort_values(ascending=False).to_frame("Total Sales"))

print("\nSales by branch:")
display(df.groupby("Branch")["Total"].sum().sort_values(ascending=False).to_frame("Total Sales"))
```

## 6. Business insights
The exact values below are produced from the dataset when the notebook is run. The points focus on sales, customers, products, payment behaviour, satisfaction and operational timing.

```python
# Useful evidence tables for writing insights
print("Top product line:", df.groupby("Product line")["Total"].sum().idxmax())
print("Lowest product line:", df.groupby("Product line")["Total"].sum().idxmin())
print("Top branch:", df.groupby("Branch")["Total"].sum().idxmax())
print("Top payment method:", df["Payment"].value_counts().idxmax())
print("Top customer type by sales:", df.groupby("Customer type")["Total"].sum().idxmax())
print("Top sales hour:", int(df.groupby("Hour").size().idxmax()))
```

### Suggested evidence-based insight list
1. Food and beverages is one of the strongest revenue-generating product lines, so it deserves close stock monitoring.
2. Health and beauty is among the weaker revenue categories, creating an opportunity for targeted promotions.
3. Branch C/Naypyitaw is the strongest branch by total sales in this dataset.
4. Sales are fairly balanced across the three branches, so the difference is meaningful but not extreme.
5. Cash, E-wallet and credit-card payments are all widely used, so the business should keep all three available.
6. Member and normal customers contribute similar levels of revenue, showing that membership has room for deeper engagement.
7. Average transaction value is about $323, while the median is much lower, showing that larger transactions pull the average upward.
8. Quantity has a clear positive relationship with total transaction value.
9. Unit price also has a positive relationship with total value, but quantity is a more direct driver of basket size.
10. Customer ratings cluster around 7/10, suggesting generally acceptable but not exceptional satisfaction.
11. The rating range is 4–10, so there are no extremely low ratings in this dataset.
12. Gross income moves closely with sales-related variables because the dataset uses a fixed gross margin percentage.
13. Transaction activity changes by hour, which can help with staff scheduling.
14. Monthly sales fluctuate across the quarter rather than increasing steadily.
15. Product lines are relatively close in sales, so the business should avoid depending on only one category.
16. High-value transactions are visible as upper-end points in the total-sales distribution and should be studied for basket composition.
17. Branch-level rating and sales should be considered together; a branch can perform well financially without having the highest rating.
18. The strongest opportunities are therefore a mix of inventory focus, loyalty engagement and time-based operational planning.

## 7. Business recommendations
1. **Protect stock for high-performing categories:** keep stronger-selling categories such as Food and Beverages and Sports and Travel well stocked because stock-outs can directly affect revenue.
2. **Target weaker categories:** run short promotions or bundle offers for Health and Beauty rather than giving blanket discounts across the store.
3. **Strengthen the membership program:** members and normal customers are close in contribution, so member-only offers should be tested to increase repeat purchases.
4. **Plan staffing around busy hours:** use the hourly transaction pattern to place more staff at peak periods and reduce idle capacity during quieter hours.
5. **Improve customer experience using ratings:** monitor branches and product lines with lower average ratings and collect simple feedback on service, availability and checkout experience.
6. **Keep payment options flexible:** payment usage is spread across cash, E-wallet and cards, so removing one option could create unnecessary friction.
7. **Study high-value baskets:** review transactions near the upper end of the Total distribution to understand which products are commonly purchased together.
